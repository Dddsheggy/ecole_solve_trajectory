```python
import pickle
import gzip
import numpy as np


sample = 'samples/sample_20368.pkl'

with gzip.open(sample, 'rb') as f:
    d = pickle.load(f)

print(d)

for data in d[2:]:
    print(data)
    print(len(data))
    cnt = 0
    
    for num in data:
        if np.isnan(num):
            cnt += 1
    print(cnt)
```
